```python
import pandas as pd

from imports import *
```

```python
from config import dir_config

raw_dir = Path(dir_config.data.raw)
processed_dir = Path(dir_config.data.processed)

filtered_data = pd.read_csv(Path(processed_dir, "processed_all_data_accu_60_filtered.csv"), index_col=None)
processed_metadata = pd.read_csv(Path(processed_dir, "processed_metadata_all_data_accu_60.csv"), index_col=None)
```

```python
filtered_data.head()
```

```python
behavior_df = pd.DataFrame()

# now grab the relevant columns from the filtered data and put them in the behavior_df
behavior_df["session_id"] = filtered_data["session_id"]
behavior_df["prior_direction"] = filtered_data["prior_direction"]
behavior_df["signed_coherence"] = filtered_data["signed_coherence"]
behavior_df["color"] = filtered_data["color"]
behavior_df["target"] = filtered_data["target"]
behavior_df["choice"] = filtered_data["choice"]
behavior_df["outcome"] = filtered_data["outcome"]
behavior_df["rt"] = filtered_data["reaction_time"]


behavior_df = behavior_df.dropna().reset_index(drop=True)
if behavior_df["signed_coherence"].max() > 1:
    print("Warning: signed_coherence has values greater than 1, dividing by 100!")
    behavior_df["signed_coherence"] /= 100  # convert to proportion for plotting
if behavior_df["rt"].max() > 10:
    print("Warning: rt has values greater than 10 seconds, dividing by 1000!")
    behavior_df["rt"] /= 1000  # convert to seconds for plotting

# sanity check
print(behavior_df["signed_coherence"].value_counts().sort_index())
print(behavior_df["target"].value_counts())
print(behavior_df["choice"].value_counts())
```

```python
behavior_df.to_csv(Path(processed_dir, "ddm", "behavior_data.csv"), index=False)
```
